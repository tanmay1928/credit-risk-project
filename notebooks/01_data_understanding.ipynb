```python
numeric_cols = [
    "age",
    "bank_account_vintage_years",
    "monthly_income_inr",
    "existing_loans_count",
    "existing_emi_monthly_inr",
    "credit_utilization_ratio_pct",
    "num_credit_inquiries_last_6m",
    "late_payments_last_12m",
    "loan_amount_requested_inr",
    "loan_tenure_months",
    "cibil_score",
    "interest_rate_offered_pct",
    "default_risk_score"
]

outlier_summary = []

for col in numeric_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = ((df[col] < lower_bound) | (df[col] > upper_bound)).sum()

    outlier_summary.append({
        "variable": col,
        "Q1": Q1,
        "Q3": Q3,
        "lower_bound": lower_bound,
        "upper_bound": upper_bound,
        "outlier_count": outliers,
        "outlier_pct": outliers / len(df) * 100
    })

outlier_summary = pd.DataFrame(outlier_summary)

outlier_summary.sort_values(
    "outlier_pct",
    ascending=False
)
```

```python
# Separate target and features

X = df.drop(columns=["default_flag", "applicant_id"])
y = df["default_flag"]

print("X shape:", X.shape)
print("y shape:", y.shape)
```

```python
print("Non-default:", (y == 0).sum())
print("Default:", (y == 1).sum())

print("\nDefault rate:", y.mean())
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training set:", X_train.shape)
print("Test set:", X_test.shape)

print("\nTraining default rate:", y_train.mean())
print("Test default rate:", y_test.mean())
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder

# Identify column types
numeric_features = X.select_dtypes(
    include=["int64", "float64"]
).columns

categorical_features = X.select_dtypes(
    include=["object"]
).columns

print("Numerical features:")
print(numeric_features.tolist())

print("\nCategorical features:")
print(categorical_features.tolist())
```

```python
# Preprocessing for numerical variables
numeric_transformer = Pipeline(
    steps=[
        ("imputer", SimpleImputer(strategy="median"))
    ]
)

# Preprocessing for categorical variables
categorical_transformer = Pipeline(
    steps=[
        ("imputer", SimpleImputer(strategy="most_frequent")),
        ("onehot", OneHotEncoder(handle_unknown="ignore"))
    ]
)

# Combine both
preprocessor = ColumnTransformer(
    transformers=[
        ("num", numeric_transformer, numeric_features),
        ("cat", categorical_transformer, categorical_features)
    ]
)
```

```python
from sklearn.linear_model import LogisticRegression

logistic_model = Pipeline(
    steps=[
        ("preprocessor", preprocessor),
        ("model", LogisticRegression(
            max_iter=1000,
            class_weight="balanced"
        ))
    ]
)

logistic_model.fit(X_train, y_train)
```

```python
y_pred = logistic_model.predict(X_test)

y_proba = logistic_model.predict_proba(X_test)[:, 1]
```

```python
from sklearn.metrics import (
    classification_report,
    confusion_matrix,
    roc_auc_score
)

print("Classification Report:\n")
print(classification_report(y_test, y_pred))

print("\nConfusion Matrix:\n")
print(confusion_matrix(y_test, y_pred))

print("\nROC-AUC:", roc_auc_score(y_test, y_proba))
```

```python
from sklearn.ensemble import RandomForestClassifier

random_forest_model = Pipeline(
    steps=[
        ("preprocessor", preprocessor),
        ("model", RandomForestClassifier(
            n_estimators=300,
            random_state=42,
            class_weight="balanced",
            n_jobs=-1
        ))
    ]
)

random_forest_model.fit(X_train, y_train)
```

```python
rf_pred = random_forest_model.predict(X_test)
rf_proba = random_forest_model.predict_proba(X_test)[:, 1]
```

```python
print("Classification Report:\n")
print(classification_report(y_test, rf_pred))

print("\nConfusion Matrix:\n")
print(confusion_matrix(y_test, rf_pred))

print("\nROC-AUC:", roc_auc_score(y_test, rf_proba))
```

```python
from sklearn.model_selection import GridSearchCV

rf_pipeline = Pipeline(
    steps=[
        ("preprocessor", preprocessor),
        ("model", RandomForestClassifier(
            random_state=42,
            class_weight="balanced",
            n_jobs=-1
        ))
    ]
)

param_grid = {
    "model__n_estimators": [200, 300],
    "model__max_depth": [None, 10, 20],
    "model__min_samples_leaf": [1, 5]
}

grid_search = GridSearchCV(
    rf_pipeline,
    param_grid=param_grid,
    scoring="roc_auc",
    cv=3,
    n_jobs=-1,
    verbose=1
)

grid_search.fit(X_train, y_train)
```

```python
print("Best parameters:")
print(grid_search.best_params_)

print("\nBest CV ROC-AUC:")
print(round(grid_search.best_score_, 3))
```

## Final Threshold Selection

The classification threshold is selected using a validation set derived from the training data. The test set is kept untouched for final model evaluation.

```python
from sklearn.model_selection import train_test_split
from sklearn.base import clone
from sklearn.metrics import f1_score
import numpy as np

# Split training data into model-fitting and validation sets
X_fit, X_val, y_fit, y_val = train_test_split(
    X_train,
    y_train,
    test_size=0.20,
    stratify=y_train,
    random_state=42
)

# Clone the tuned model so the original remains unchanged
threshold_model = clone(grid_search.best_estimator_)

# Train on the fitting portion
threshold_model.fit(X_fit, y_fit)

# Get validation probabilities
val_proba = threshold_model.predict_proba(X_val)[:, 1]

# Find the threshold that gives the highest F1 score
thresholds = np.arange(0.10, 0.81, 0.01)

f1_scores = []

for threshold in thresholds:
    val_pred = (val_proba >= threshold).astype(int)
    f1_scores.append(f1_score(y_val, val_pred))

best_threshold = thresholds[np.argmax(f1_scores)]
best_f1 = max(f1_scores)

print(f"Best threshold: {best_threshold:.2f}")
print(f"Validation F1-score: {best_f1:.3f}")
```

```python
from sklearn.metrics import (
    roc_auc_score,
    classification_report,
    confusion_matrix
)

# Train the final tuned model on the complete training set
final_model = clone(grid_search.best_estimator_)
final_model.fit(X_train, y_train)

# Final predictions on untouched test data
test_proba = final_model.predict_proba(X_test)[:, 1]
test_pred = (test_proba >= best_threshold).astype(int)

# Final metrics
final_auc = roc_auc_score(y_test, test_proba)

print(f"Test ROC-AUC: {final_auc:.3f}")
print(f"Final threshold: {best_threshold:.2f}")
print("\nClassification Report:")
print(classification_report(y_test, test_pred))

print("Confusion Matrix:")
print(confusion_matrix(y_test, test_pred))
```

## Final Risk Segmentation

Applicants are grouped into three exploratory risk bands based on their predicted probability of default. These bands are used for business interpretation and are not calibrated probability-of-default estimates.

```python
# Create final prediction dataset
final_risk = X_test.copy()

final_risk["actual_default"] = y_test.values
final_risk["default_probability"] = test_proba

# Create exploratory risk bands
final_risk["risk_band"] = pd.cut(
    final_risk["default_probability"],
    bins=[0, 0.20, 0.50, 1.00],
    labels=["Low Risk", "Medium Risk", "High Risk"],
    include_lowest=True
)

# Summarise each risk band
risk_summary = (
    final_risk
    .groupby("risk_band", observed=False)
    .agg(
        applicants=("actual_default", "count"),
        actual_defaults=("actual_default", "sum"),
        average_predicted_probability=("default_probability", "mean"),
        actual_default_rate=("actual_default", "mean")
    )
    .reset_index()
)

risk_summary["average_predicted_probability"] *= 100
risk_summary["actual_default_rate"] *= 100

risk_summary
```

```python
# Overall test-set performance
overall_default_rate = final_risk["actual_default"].mean() * 100

print(f"Total test applicants: {len(final_risk):,}")
print(f"Actual defaults: {final_risk['actual_default'].sum():,}")
print(f"Overall default rate: {overall_default_rate:.2f}%")
```
